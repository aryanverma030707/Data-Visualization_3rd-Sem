Parameters vs. Arguments

These two terms are often used interchangeably, but they mean different things — and knowing the distinction will make you sound (and think) like a precise programmer.

Parameter: The variable name listed in the function's definition. It's a placeholder.
Argument: The actual value you pass into the function when you call it.

****Argument Types**

Python offers several flexible ways to pass arguments into a function. Understanding each type lets you design functions that are both powerful and easy to use.**


**Positional Arguments  =  
The most common type — arguments are matched to parameters based on their order.**


```python
def describe_pet(animal, name):
    print(f"{name} is a {animal}.")

describe_pet("dog", "Rex")   # animal="dog", name="Rex"
```

****Variable-Length Arguments: *args and*** **kwargs**

Sometimes you don't know in advance how many arguments will be passed. Python handles this with two special syntaxes.

***args** (non-keyword variable arguments) Collects any number of positional arguments into a tuple.

****kwargs** (keyword variable arguments) Collects any number of keyword arguments into a dictionary.

Keyword arguments are arguments you pass to a function by explicitly naming the parameter they belong to, using parameter=value syntax — instead of relying on position/order.

```python
def add(**kwargs):
    return sum(kwargs.values())

data = {}
data["a"] = int(input("Enter first number: "))
data["b"] = int(input("Enter second number: "))

print(add(**data))  # unpacks dict as keyword arguments
```

```python
def add(**kwargs):
    print(type(kwargs))   # <class 'dict'>
    print(kwargs)          # e.g. {'a': 10, 'b': 20, 'c': 30}
    total = 0
    for value in kwargs.values():
        total += value
    return total

print(add(a=10, b=20, c=30))  # Output: 60
```

```python
def add(**kwargs):
    total = 0
    for key, value in kwargs.items():
        print(key, "->", value)
        total += value
    return total

print(add(a=10, b=20, c=30))
```

```python
def add(**kwargs):
    total = 0
    for key in kwargs.keys():
        print(key, "->", kwargs[key])
        total += kwargs[key]
    return total

print(add(a=10, b=20, c=30))
```

```python
def division(a, b):
    if b != 0:
        return (a / b, a // b)
    else:
        print("Error: Cannot divide by zero.")
        return None

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

result = division(a, b)
print(type(result))
if result is not None:
    print(f"True division: {result[0]}, Floor division: {result[1]}")
```

```python
def is_perfect(num):
    """Return True if num is a perfect number (optimized version)."""
    if num < 2:
        return False
    
    divisor_sum = 1  # 1 is always a proper divisor
    i = 2
    while i * i <= num:
        if num % i == 0:
            divisor_sum += i
            partner = num // i
            if partner != i:
                divisor_sum += partner
        i += 1
    
    return divisor_sum == num


def find_perfect_numbers(start, end):
    return [num for num in range(start, end + 1) if is_perfect(num)]


print("Perfect numbers between 1 and 1000:", find_perfect_numbers(1, 1000))
```

```python
t = (10, 20, 30, 40, 50)

print(t[1:4])     # (20, 30, 40)
print(t[:3])      # (10, 20, 30)
print(t[2:])      # (30, 40, 50)
print(t[:])       # (10, 20, 30, 40, 50)
print(t[::2])     # (10, 30, 50)
print(t[1::2])    # (20, 40)
print(t[::-1])    # (50, 40, 30, 20, 10)
```
