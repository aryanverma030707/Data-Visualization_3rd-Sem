```python
# List, tuples, sets, dictionaries, strings, numbers, boolean, NoneType
```

```python
# List is heterogeneous and ordered collection of data. It is mutable and allows duplicate elements.
# It is defined using square brackets [].
my_list = [10, "Hello", 3.14, True]
print(my_list, type(my_list))
print(my_list[0])  # Accessing first element
my_list[1] = "World"  # Modifying second element
print(my_list)
```

```python
#Tuple is heterogeneous and ordered collection of data. It is immutable and allows duplicate elements.
# It is defined using parentheses ().
my_tuple = (10, "Hello", 3.14, True)
print(my_tuple,type(my_tuple))
print(my_tuple[0]) # Accessing first element
print(my_tuple)
```

```python
#set is heterogeneous and unordered collection of data. It is mutable and does not allow duplicate elements.
# It is defined using curly braces {}.
my_set = {10, "Hello", 3.14, True}
print(my_set, type(my_set)) 
for item in my_set:
    print(item)
```

```python
# Dictionary is heterogeneous and ordered collection of data. It is mutable and does not allow duplicate keys.
# It is defined using curly braces {} with key-value pairs.
my_dict = {"name": "Aryan", "age": 24, "is_student": True}
print(my_dict, type(my_dict))
print(my_dict["name"])  # Accessing value by key   
print(my_dict.get("age"))  # Accessing value by key using get() method 
```

# Type Casting / Type Conversion 
# 2 types=
# Implicit Type Casting and Explicit Type Casting 

```python
a=10 # int 
b=3 # int
c=a/b # system automatically converts the result to float {Implicit type conversion}
d=True
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))
```

```python
a=10 # int 
b=3 # int
c=a/b # system automatically converts the result to float {Implicit type conversion}
c=int(c) # Explicit type conversion
print(a, type(a))
print(b, type(b))
print(c, type(c))
```

```python
age = 24.0
age_str = str(age) # Explicit type conversion from float to string
print(age_str, type(age_str))
print(int(float(age_str)), type(int(float(age_str))))
```

```python
x="10.324"

y = int(float(x)) # Explicit type conversion from string to int
print(y, type(y))
y = float(x) # Explicit type conversion from string to float
print(y, type(y))
```

```python
num1 = eval(input("Enter first number: "))
num2 = eval(input("Enter second number: "))
sum_result = num1 + num2
difference_result = num1 - num2
product_result = num1 * num2
quotient_result = num1 / num2
power_result = num1 ** num2
print("The sum of", num1, "and", num2, "is:", sum_result)
print("The difference of", num1, "and", num2, "is:", difference_result)
print("The product of", num1, "and", num2, "is:", product_result)
print("The quotient of", num1, "and", num2, "is:", quotient_result)
print("Power:", power_result)
```

```python
#Bitwise Operators
# { & | ~ ^ >> << } 
a=100
b=50
print(a & b)  # Bitwise AND
print(a | b)  # Bitwise OR
print(~a)     # Bitwise NOT
print(a ^ b)  # Bitwise XOR 
print(a << 2) # Bitwise left shift
print(a >> 2) # Bitwise right shift
```

```python
#Assignment Operators
# { =, +=, -=, *=, /=, %=, //=, **=  &=, |=, ^=, >>=, <<= }
a=eval(input("Enter a number: ")) # if Input is 10
a+=5
print(a)  # Output: 15
a -= 3
print(a)  # Output: 12
a*=2
print(a)  # Output: 24  
a /=2
print(a)  # Output:12
a //= 5
print(a)  # Output:2
a **= 5
print(a)  # Output: 32
```
