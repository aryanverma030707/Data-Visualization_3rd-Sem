```python
# List, tuples, sets, dictionaries, strings, numbers, boolean, NoneType
```

```python
# List is heterogeneous and ordered collection of data. It is mutable and allows duplicate elements.
# It is defined using square brackets [].
my_list = [10, "Hello", 3.14, True]
print(my_list, type(my_list))
print(my_list[0])  # Accessing first element
my_list[1] = "World"  # Modifying second element
print(my_list)
```

```python
#Tuple is heterogeneous and ordered collection of data. It is immutable and allows duplicate elements.
# It is defined using parentheses ().
my_tuple = (10, "Hello", 3.14, True)
print(my_tuple,type(my_tuple))
print(my_tuple[0]) # Accessing first element
print(my_tuple)
```

```python
#set is heterogeneous and unordered collection of data. It is mutable and does not allow duplicate elements.
# It is defined using curly braces {}.
my_set = {10, "Hello", 3.14, True}
print(my_set, type(my_set)) 
for item in my_set:
    print(item)
```

```python
# Dictionary is heterogeneous and unordered collection of data. It is mutable and does not allow duplicate keys.
# It is defined using curly braces {} with key-value pairs.
my_dict = {"name": "Aryan", "age": 24, "is_student": True}
print(my_dict, type(my_dict))
print(my_dict["name"])  # Accessing value by key   
print(my_dict.get("age"))  # Accessing value by key using get() method 
```
